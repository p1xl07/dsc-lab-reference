# 04 — Model Evaluation Quick Reference

## Regression metrics

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_true = np.array([3, -0.5, 2, 7])
y_pred = np.array([2.5, 0.0, 2, 8])

mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_true, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2:", r2)
```

### Formulas
- `MAE = mean(abs(y - y_hat))`
- `MSE = mean((y - y_hat)^2)`
- `RMSE = sqrt(MSE)`
- `R² = 1 - (sum((y-y_hat)^2) / sum((y-y_mean)^2))`

Lower MAE/MSE/RMSE is better. R² closer to 1 generally indicates a better fit; it can be negative.

## Classification metrics

```python
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score, classification_report

y_true = [0, 1, 1, 0, 1, 0]
y_pred = [0, 1, 0, 0, 1, 1]

print("Confusion matrix:\n", confusion_matrix(y_true, y_pred))
print("Accuracy:", accuracy_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred))
print("Recall:", recall_score(y_true, y_pred))
print("F1:", f1_score(y_true, y_pred))
print(classification_report(y_true, y_pred))
```

### Confusion matrix (binary classification)
- TP: actual 1, predicted 1
- TN: actual 0, predicted 0
- FP: actual 0, predicted 1
- FN: actual 1, predicted 0

Formulas:
- `Accuracy = (TP + TN) / (TP + TN + FP + FN)`
- `Precision = TP / (TP + FP)`
- `Recall = TP / (TP + FN)`
- `F1 = 2 * Precision * Recall / (Precision + Recall)`

In scikit-learn's binary metrics, class `1` is treated as the positive class by default.
