# 02 — Data Preprocessing Quick Reference

Common operations using Pandas and scikit-learn.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler
```

## Read and inspect data

```python
df = pd.read_csv("data.csv")
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

## Handle missing values

```python
# Drop rows containing missing values
df_drop = df.dropna()

# Fill numeric missing values with the column mean
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill categorical missing values with the most frequent value
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

## Encode categorical data

```python
# Label encoding: usually for target labels / ordinal categories
le = LabelEncoder()
df["Target"] = le.fit_transform(df["Target"])

# One-hot encoding: nominal input categories
df = pd.get_dummies(df, columns=["City"], dtype=int)
```

## Important reminders
- `axis=0` means rows; `axis=1` means columns.
- `drop("column", axis=1)` removes a column.
- Fit scalers on training data only in a proper ML workflow; transform test data with the same scaler.
- `get_dummies()` is a quick way to one-hot encode columns.
- Use `random_state` to make train/test splits reproducible.
