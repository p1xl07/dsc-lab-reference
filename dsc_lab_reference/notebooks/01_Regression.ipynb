# 01 — Regression Quick Reference

Covers simple and multiple linear regression (manual NumPy) and single-/multiple-feature logistic regression (scikit-learn).

**Imports**

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
```

## A. Simple Linear Regression

Model: `y_hat = b0 + b1*x`

Slope: `b1 = sum((x-x_mean)*(y-y_mean)) / sum((x-x_mean)**2)`

Intercept: `b0 = y_mean - b1*x_mean`

Mean squared error: `MSE = mean((y-y_hat)**2)`

```python
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4, 5], dtype=float)

x_mean = np.mean(x)
y_mean = np.mean(y)
b1 = np.sum((x-x_mean)*(y-y_mean)) / np.sum((x-x_mean)**2)
b0 = y_mean - b1*x_mean

y_pred = b0 + b1*x
print("b0:", b0, "b1:", b1)
print("MSE:", np.mean((y-y_pred)**2))

x_new = float(input("Enter x: "))
print("Prediction:", b0 + b1*x_new)
```

## B. Multiple Linear Regression

Model: `y_hat = b0 + b1*x1 + b2*x2 + ... + bp*xp`

Matrix solution: `b = (X.T @ X)^(-1) @ X.T @ y`

Use the pseudoinverse in code: `b = pinv(X_with_intercept) @ y`.

```python
X = np.array([[1, 2], [2, 1], [3, 4], [4, 3], [5, 5]], dtype=float)
y = np.array([5, 6, 13, 14, 20], dtype=float)

X_b = np.c_[np.ones(X.shape[0]), X]
b = np.linalg.pinv(X_b) @ y
y_pred = X_b @ b

print("Coefficients [b0, b1, b2]:", b)
print("MSE:", np.mean((y-y_pred)**2))

values = [float(input(f"Enter feature {i+1}: ")) for i in range(X.shape[1])]
new_row = np.array([1] + values)
print("Prediction:", new_row @ b)
```

## C. Logistic Regression — Single Feature

Sigmoid: `p = 1 / (1 + exp(-z))`, where `z = b0 + b1*x`.

For a 0.5 threshold: predict 1 if `p >= 0.5`, otherwise 0.

Logistic regression is for classification. `X` must be 2D, even with one feature.

```python
X = np.array([[1], [2], [3], [4], [5], [6]], dtype=float)
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

x_new = float(input("Enter x: "))
print("Probability of class 1:", model.predict_proba([[x_new]])[0][1])
print("Prediction:", model.predict([[x_new]])[0])
```

## D. Logistic Regression — Multiple Features

`z = b0 + b1*x1 + b2*x2 + ... + bp*xp`; `p = sigmoid(z)`.

Only the number of columns in `X` changes compared with single-feature logistic regression.

```python
X = np.array([
    [1,  2], [2, 1], [2, 3], [4, 2],
    [5,  4], [6, 5], [7, 4], [8, 6]
], dtype=float)
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

values = [float(input(f"Enter feature {i+1}: ")) for i in range(X.shape[1])]
new_row = np.array([values])
print("Probability of class 1:", model.predict_proba(new_row)[0][1])
print("Prediction:", model.predict(new_row)[0])
```

## Remember
- Linear regression here is calculated manually; no `.fit()` is used.
- Logistic regression uses `model.fit(X, y)`.
- For multiple regression, rows are observations and columns are features.
- Add a ones column to linear-regression `X` to calculate the intercept.
- Keep prediction feature order exactly the same as training feature order.
- `predict()` gives a class; `predict_proba()` gives probabilities.
