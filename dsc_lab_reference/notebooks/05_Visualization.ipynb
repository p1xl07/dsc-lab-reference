# 05 — Data Visualization Quick Reference

Using Matplotlib and Seaborn.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [35, 40, 55, 65, 80],
    "Group": ["A", "A", "B", "B", "B"]
})
```

## Line plot

```python
plt.plot(df["Hours"], df["Marks"], marker="o")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.title("Hours vs Marks")
plt.show()
```

## Scatter plot

```python
plt.scatter(df["Hours"], df["Marks"])
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.title("Scatter Plot")
plt.show()
```

## Bar chart and histogram

```python
plt.bar(df["Group"], df["Marks"])
plt.xlabel("Group")
plt.ylabel("Marks")
plt.title("Marks by Group")
plt.show()

plt.hist(df["Marks"], bins=5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```

## Box plot and heatmap

```python
sns.boxplot(x=df["Marks"])
plt.show()

numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

## Remember
- `plt.plot()` — line graph.
- `plt.scatter()` — relationship between two numerical variables.
- `plt.bar()` — compare categories.
- `plt.hist()` — distribution of one numerical variable.
- `sns.boxplot()` — spread and possible outliers.
- `sns.heatmap()` — visualize a matrix, commonly correlations.
- Add labels/title and call `plt.show()`.
