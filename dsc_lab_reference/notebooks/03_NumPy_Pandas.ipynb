# 03 — NumPy and Pandas Quick Reference

## NumPy

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([[1, 2], [3, 4]])

print(a.shape, b.shape)
print(a.ndim, b.ndim)
print(np.zeros((2, 3)))
print(np.ones((2, 2)))
print(np.arange(1, 10, 2))
print(np.mean(a), np.median(a), np.std(a), np.sum(a))
print(np.min(a), np.max(a))
```

## Array indexing and operations

```python
a = np.array([10, 20, 30, 40])
print(a[0])
print(a[1:3])
print(a[a > 20])
print(a + 5)
print(a * 2)

m = np.array([[1, 2], [3, 4]])
print(m[0, 1])
print(m[:, 0])  # first column
print(m[1, :])  # second row
print(m.T)      # transpose
```

## Pandas

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Mina"],
    "Age": [20, 21, 19],
    "Marks": [85, 72, 91]
})

print(df.head())
print(df["Age"])
print(df[["Name", "Marks"]])
print(df.loc[0])                 # label-based row
print(df.iloc[0])                # position-based row
print(df[df["Marks"] > 80])
```

## Common DataFrame operations

```python
print(df["Marks"].mean())
print(df.sort_values("Marks", ascending=False))
print(df["Age"].value_counts())
print(df.isnull().sum())
df["Passed"] = df["Marks"] >= 40
df = df.drop(columns=["Passed"])
df.to_csv("output.csv", index=False)
```

## Quick distinctions
- NumPy array: numerical data and fast mathematical operations.
- Pandas Series: one labelled column.
- Pandas DataFrame: tabular data with rows and columns.
- `loc` uses labels; `iloc` uses integer positions.
- `df.shape` returns `(rows, columns)`.
- `df.columns` lists column names.
